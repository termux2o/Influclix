```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import csv
import os
from webdriver_manager.chrome import ChromeDriverManager

# LinkedIn Credentials
USERNAME = "aakash@brozmedia.co.in"
PASSWORD = "@1234Aakash"

# LinkedIn Search URL
SEARCH_URL = "https://www.linkedin.com/search/results/content/?keywords=%22Hiring%20website%22%20OR%20%22hiring%20web%20developer%22%20OR%20%22looking%20for%20website%20designer%22%20OR%20%22wordpress%20developer%22&origin=GLOBAL_SEARCH_HEADER&sid=sX0&sortBy=%22date_posted%22"

# Output CSV File
OUTPUT_FILE = "linkedin_profiles_filtered.csv"

# Initialize WebDriver
options = webdriver.ChromeOptions()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# Open LinkedIn login page
driver.get("https://www.linkedin.com/login")
time.sleep(3)

# Login to LinkedIn
driver.find_element(By.ID, "username").send_keys(USERNAME)
driver.find_element(By.ID, "password").send_keys(PASSWORD)
driver.find_element(By.XPATH, '//button[@type="submit"]').click()
time.sleep(5)

# Navigate to search results page
driver.get(SEARCH_URL)
time.sleep(5)

print("🔍 Searching for LinkedIn leads...")

# Load existing records into a set
def load_existing_data():
    existing_records = set()
    if os.path.exists(OUTPUT_FILE):
        with open(OUTPUT_FILE, "r", encoding="utf-8") as file:
            reader = csv.reader(file)
            next(reader, None)  # Skip header row if present
            for row in reader:
                if len(row) >= 2:  # Ensure valid entries
                    existing_records.add((row[0], row[1]))  # Store name-link pairs
    return existing_records

# Extract profiles and append only new records
def extract_and_append(existing_records):
    profiles = []
    listings = driver.find_elements(By.XPATH, "//div[contains(@class, 'update-components-actor')]")

    for listing in listings:
        try:
            name_element = listing.find_element(By.XPATH, ".//a[contains(@class, 'update-components-actor__meta-link')]")
            name = name_element.text.strip()
            profile_link = name_element.get_attribute("href")

            # Remove unnecessary details like "3rd+", "2nd+", job titles
            name = name.split("\n")[0]  # Keep only the name, removing extra text if any

            if (name, profile_link) not in existing_records:
                profiles.append((name, profile_link))
                existing_records.add((name, profile_link))  # Update existing records

        except:
            continue  # Skip if extraction fails

    # Write new profiles to CSV in a proper row format
    if profiles:
        file_exists = os.path.exists(OUTPUT_FILE)
        with open(OUTPUT_FILE, "a", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)

            # Write headers if file is new
            if not file_exists:
                writer.writerow(["Name", "Profile Link"])

            writer.writerows(profiles)  # Write each row properly
            print(f"✅ {len(profiles)} new entries added!")

# Scroll and load more results dynamically
def scroll_and_load():
    existing_records = load_existing_data()
    
    while True:
        driver.find_element(By.TAG_NAME, 'body').send_keys(Keys.END)
        time.sleep(3)  # Allow time for content to load

        extract_and_append(existing_records)  # Extract and save data

        try:
            show_more_button = WebDriverWait(driver, 5).until(
                EC.element_to_be_clickable((By.XPATH, '//button[contains(@class, "scaffold-finite-scroll__load-button")]'))
            )
            if show_more_button.is_displayed():
                show_more_button.click()
                print("🔄 Clicked 'Show more results' button!")
                time.sleep(3)
        except:
            print("✅ No more 'Show more results' button. Scrolling finished.")
            break  # Exit when no more results

# Start scrolling and extracting in real-time
scroll_and_load()

print(f"🚀 Scraping complete. Data saved in '{OUTPUT_FILE}'.")

# Close the browser
driver.quit()
```
