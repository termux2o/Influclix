```python
import instaloader
import pandas as pd

# Load usernames from CSV file
csv_file = "usernamse.csv"

# Read the CSV file
try:
    df = pd.read_csv(csv_file)
except FileNotFoundError:
    print(f"Error: File '{csv_file}' not found.")
    exit()

# Ensure 'username' column exists
if 'Username' not in df.columns:
    print("Error: 'username' column missing in CSV.")
    exit()

# Initialize Instaloader (no login required)
ig = instaloader.Instaloader()

# Create a new column for followers
df["followers"] = None

# Fetch followers count for each username
for index, row in df.iterrows():
    usrname = row["Username"]
    
    try:
        profile = instaloader.Profile.from_username(ig.context, usrname)
        df.at[index, "followers"] = profile.followers
        print(f"Fetched: {usrname} - {profile.followers} followers")
    
    except instaloader.exceptions.ProfileNotExistsException:
        print(f"Error: Profile '{usrname}' does not exist or is private.")
        df.at[index, "followers"] = "Not Found"
    
    except instaloader.exceptions.ConnectionException:
        print("Error: Connection issue, try again later.")
        df.at[index, "followers"] = "Error"

# Save updated CSV
df.to_csv(csv_file, index=False)
print(f"\n✅ Updated '{csv_file}' with followers count.")
```

```python
import pandas as pd
import csv
import time
import os
from instagrapi import Client

# Instagram credentials
USERNAME = "aakash_sharma_413"
PASSWORD = "@1234Aakash"  # Replace with your actual password

# Initialize the Client
cl = Client()

# Function to login with retries
def login_instagram():
    attempts = 3  # Retry login up to 3 times
    for attempt in range(attempts):
        try:
            cl.login(USERNAME, PASSWORD)
            print("✅ Login successful!")
            return True
        except Exception as e:
            print(f"❌ Login attempt {attempt+1} failed: {e}")
            if "challenge_required" in str(e):
                challenge_url = cl.challenge_resolve(cl.last_json)
                print(f"🔗 Challenge required! Please visit: {challenge_url}")
                input("✅ Solve the challenge and press Enter to continue...")
                continue  # Retry login after solving challenge
            time.sleep(10)  # Wait before retrying
    return False

# Attempt login
if not login_instagram():
    print("❌ Failed to login after multiple attempts. Exiting...")
    exit()

# Read the input CSV file
input_file = 'usernamse.csv'
df = pd.read_csv(input_file)

# Output CSV file to store results
output_file = 'updated_follower2.csv'

# Check if output file exists
file_exists = os.path.exists(output_file)

# Load existing processed data
if file_exists:
    processed_df = pd.read_csv(output_file)
    processed_users = dict(zip(processed_df['Username'], processed_df['Followers']))
else:
    processed_users = {}

# Open output CSV in append mode
with open(output_file, 'a', newline='') as file:
    csv_writer = csv.writer(file)

    # Write the header if file is new
    if not file_exists:
        csv_writer.writerow(['Username', 'Followers'])

    # Process each username
    for index, row in df.iterrows():
        usrname = row['Username']

        # Skip usernames that already have follower counts
        if usrname in processed_users and not pd.isna(processed_users[usrname]):
            print(f"⏩ Skipping {usrname} (Already fetched: {processed_users[usrname]} followers)")
            continue

        retry_attempts = 3  # Retry fetching data up to 3 times
        login_attempts = 0  # Track login attempts for this user

        for attempt in range(retry_attempts):
            try:
                # Fetch followers count
                user_id = cl.user_id_from_username(usrname)
                user_info = cl.user_info(user_id)
                followers = user_info.follower_count

                # Print and store data
                print(f"✅ Username: {usrname}, Followers: {followers}")
                csv_writer.writerow([usrname, followers])

                # Save progress
                file.flush()
                break  # Exit retry loop after success

            except Exception as e:
                print(f"❌ Error fetching {usrname} (Attempt {attempt+1}): {e}")

                # Handle login_required error
                if "login_required" in str(e):
                    login_attempts += 1
                    if login_attempts >= 2:  # Skip after 2 failed login attempts
                        print(f"❌ Skipping {usrname} after {login_attempts} failed login attempts.")
                        break  # Skip user and move to the next one
                    print("🔄 Re-logging in...")
                    if login_instagram():
                        continue  # Retry after successful re-login
                    else:
                        print("❌ Login failed, skipping user.")
                        break  # Skip user after failed re-login

                # Handle challenge required error
                if "challenge_required" in str(e):
                    challenge_url = cl.challenge_resolve(cl.last_json)
                    print(f"🔗 Challenge required for {usrname}! Please visit: {challenge_url}")
                    input("✅ Solve the challenge and press Enter to continue...")
                    if login_instagram():
                        continue  # Retry after solving challenge
                    else:
                        print(f"❌ Challenge failed for {usrname}, skipping...")
                        break  # Skip user after failed challenge resolution

                # Wait before retrying
                time.sleep(5)

print(f"\n✅ Successfully updated '{output_file}' with follower counts.")
```
