```python
import os
import csv
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager

# ✅ Setup WebDriver
chrome_options = Options()
chrome_options.add_argument("--start-maximized")
chrome_options.add_argument("--disable-blink-features=AutomationControlled")

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=chrome_options)

q = input("Enter the search query: ")

# Construct the search URL with the given query
search_url = f"https://www.google.com/search?q={q}&num=100"
driver.get(search_url)
time.sleep(3)

# ✅ Output File Setup
output_file = "25march.csv"
if not os.path.exists(output_file):
    with open(output_file, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["Username", "Bio", "Details", "Post Link"])  # Updated Column Headers

count = 0

while True:
    # ✅ Auto Scroll to Load More Results
    for _ in range(40):
        driver.execute_script("window.scrollBy(0, 500);")
        time.sleep(1)

    # ✅ Extract Search Results
    search_results = driver.find_elements(By.CSS_SELECTOR, "div.tF2Cxc")
    new_entries = []

    for result in search_results:
        try:
            # ✅ Extract Username from span.VuuXrf
            try:
                username_element = result.find_element(By.CSS_SELECTOR, "span.VuuXrf")
                username = username_element.text.replace("Instagram · ", "").strip()  # Extract username
            except:
                username = "N/A"
           # ✅ Extract Bio from h3 Tag
            Bio = result.find_element(By.CSS_SELECTOR, "h3.LC20lb.MBeuO.DKV0Md").text.strip()

            # ✅ Extract Data from First Type of Div Tag (Followers/Likes)
            try:
                followers_likes = result.find_element(By.CSS_SELECTOR, "div.VwiC3b.yXK7lf.p4wth.r025kc.hJNv6b.Hdw6tb span").get_attribute("innerText").strip()
            except:
                followers_likes = "N/A"  # Handle cases where data is missing

            # ✅ Extract Post Link from Anchor Tag
            try:
                post_link = result.find_element(By.CSS_SELECTOR, "a.zReHs").get_attribute("href").strip()
            except:
                post_link = "N/A"  # Handle cases where link is missing

            # ✅ Print and Save the Result
            print(f"username:{username} |Bio: {Bio} | Followers/Likes: {followers_likes} | Post Link: {post_link}")
            new_entries.append([username,Bio, followers_likes, post_link])
            count += 1

        except Exception as e:
            print(f"❌ Error extracting data: {e}")

    # ✅ Save Data to CSV after each iteration
    if new_entries:
        with open(output_file, "a", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerows(new_entries)
        print(f"✅ {len(new_entries)} new entries saved!")

    # ✅ Try to Click "Next Page" Automatically
    try:
        next_button = driver.find_element(By.CSS_SELECTOR, "a#pnnext")
        driver.execute_script("arguments[0].click();", next_button)
        print("➡️ Moving to the next page...")
        time.sleep(5)
    except Exception:
        print("❌ No more pages available. Exiting.")
        break

# ✅ Close WebDriver
driver.quit()
```

```python
import os
import csv
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager

# ✅ Setup WebDriver
chrome_options = Options()
chrome_options.add_argument("--start-maximized")
chrome_options.add_argument("--disable-blink-features=AutomationControlled")
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=chrome_options)

q = input("Enter the search query: ")

# ✅ Define Static Google Search URL
search_url = f'https://www.google.com/search?q={q}&num=100'
driver.get(search_url)
time.sleep(3)

# ✅ Output File Setup
output_file = "25march.csv"
if not os.path.exists(output_file):
    with open(output_file, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["Bio", "Post Link", "Details", "Username"])  # Header

count = 0

while True:
    # ✅ Auto Scroll to Load More Results
    for _ in range(40):
        driver.execute_script("window.scrollBy(0, 500);")
        time.sleep(1)

    # ✅ Extract Both Text & Video-Based Results
    search_results = driver.find_elements(By.CSS_SELECTOR, "div.tF2Cxc, div.MjjYud")
    new_entries = []

    for result in search_results:
        try:
            # ✅ Extract Bio from h3 Tag (if available)
            try:
                bio_element = result.find_element(By.CSS_SELECTOR, "h3")
                bio = bio_element.text.strip()
            except:
                bio = "N/A"

            # ✅ Extract Post Link (if available)
            try:
                post_link_element = result.find_element(By.CSS_SELECTOR, "a")
                post_link = post_link_element.get_attribute("href").strip()
            except:
                post_link = "N/A"

            # ✅ Extract Details from <div class="ITZIwc p4wth"> (if available)
            try:
                detail_element = result.find_element(By.CSS_SELECTOR, "div.ITZIwc.p4wth")
                details = detail_element.text.strip()
            except:
                details = "N/A"

            # ✅ Extract Username from span.VuuXrf (if available)
            try:
                username_element = result.find_element(By.CSS_SELECTOR, "span.VuuXrf")
                username = username_element.text.replace("Instagram · ", "").strip()
            except:
                username = "N/A"

            # ✅ Store Extracted Data
            new_entries.append((username,bio,details,post_link))
            count += 1

        except Exception as e:
            print(f"❌ Error extracting data: {e}")

    # ✅ Save Data to CSV After Each Iteration
    if new_entries:
        with open(output_file, "a", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerows(new_entries)
        print(f"✅ {len(new_entries)} new entries saved! Total: {count}")

    # ✅ Try to Click "Next Page" Automatically
    try:
        next_button = driver.find_element(By.CSS_SELECTOR, "a#pnnext")
        driver.execute_script("arguments[0].click();", next_button)
        print("➡️ Moving to the next page...")
        time.sleep(5)
    except Exception:
        print("❌ No more pages available. Exiting.")
        break

# ✅ Close WebDriver
driver.quit()
```

### duplicates drop

```python
import pandas as pd

# Load the CSV file
file_path = "25march.csv" # Update with your actual file path
df = pd.read_csv(file_path)
# Check for duplicate Username
duplicates = df.duplicated(subset=['Username'])  # Check for duplicates in 'username' column
total_duplicates = duplicates.sum()
print(f"Total duplicate Username found: {total_duplicates}")

# Remove duplicates
df_cleaned = df.drop_duplicates(subset=['Username'], keep='first')

# Print total records after removing duplicates
print(f"Total records after removing duplicates: {len(df_cleaned)}")

# Save the cleaned data back to the same file
df_cleaned.to_csv(file_path, index=False)

print("Duplicates removed and file updated successfully.")
```

```python
import re  # Import the re module

def extract_numbers(text):
    # Regular expression to match Indian phone numbers (+91 or 91 or 10-digit numbers)
    pattern = r'(\+91[\s-]?[6-9]\d{9}|91[\s-]?[6-9]\d{9}|[6-9]\d{9})'
    numbers = re.findall(pattern, str(text))
    return numbers if numbers else None

# Extract and print contact numbers from the 'Username' column
df['Contact Numbers2'] = df['Details'].apply(extract_numbers)

# Print the rows with extracted numbers
print(df[['Contact Numbers2', 'Details']].dropna())
```

```python
import os
# Function to clean and keep only 10-digit numbers
def clean_number(number):
    # Find all 10-digit numbers in the string
    digits = re.findall(r'\b\d{10}\b', str(number))
    return digits[0] if digits else None

# Apply the cleaning function to the "Contact Numbers" column
d2['Contact Numbers'] = d2['Contact Numbers'].apply(clean_number)

# Drop rows with empty or invalid numbers
d2 = d2.dropna(subset=['Contact Numbers'])

# Remove duplicate contact numbers
d2 = d2.drop_duplicates(subset=['Contact Numbers'])

# Check if the file already exists
if os.path.isfile("18march.csv"):
    # Load the existing file to avoid duplicate appending
    existing_data = pd.read_csv("18march.csv")
    # Concatenate the new and existing data, then remove duplicates
    combined_data = pd.concat([existing_data, d2]).drop_duplicates(subset=['Contact Numbers'])
    # Append to the CSV file
    combined_data.to_csv("18march.csv", index=False)
else:
    # Save the cleaned and unique data to a new CSV file if it doesn't exist
    d2.to_csv("18march.csv", index=False)

print("Cleaned and unique contacts appended to '18march.csv'")
```
