```python
import pandas as pd
from instagrapi import Client

# Instagram login
cl = Client()
cl.login("aakash_sharma_413", "@1234Aakash")

# Load usernames from CSV file
csv_file = "company_users.csv"
df = pd.read_csv(csv_file)  # Read CSV file

# Assuming the column containing usernames is named 'username'
usernames = df["username"].dropna().tolist()  # Remove any empty values

# Message to send
message = "Hello! This is an automated message for testing By Aakash sharma."

# Send message to each user
for username in usernames:
    try:
        user_id = cl.user_id_from_username(username)  # Fetch user ID
        cl.direct_send(message, [user_id])  # Send DM
        print(f"Message sent to {username}")
    except Exception as e:
        print(f"Failed to send message to {username}: {e}")
```

```python
import pandas as pd
import time
import os
from instagrapi import Client

# Instagram login credentials (Replace with your valid credentials)
INSTAGRAM_USERNAME = "aakash_sharma_413"
INSTAGRAM_PASSWORD = "@1234Aakash"

# Initialize Instagram Client
cl = Client()
cl.login(INSTAGRAM_USERNAME, INSTAGRAM_PASSWORD)

# Input and output file paths
input_file = "influencers_data.csv"
output_file = "details.csv"

# Load usernames from input CSV
df = pd.read_csv(input_file)

# Ensure 'username' column exists
if "username" not in df.columns:
    print("Error: 'username' column is missing in influencers_data.csv")
    exit()

# Remove @ from usernames and drop NaN values
df["username"] = df["username"].astype(str).str.replace("@", "").str.strip()
usernames = df["username"].dropna().unique()

# Ensure output file exists with correct headers
if not os.path.exists(output_file):
    df_output = pd.DataFrame(columns=["username", "user_id", "full_name", "followers", "following", "bio", "profile_pic_url"])
    df_output.to_csv(output_file, index=False)

# Process each username
for index, username in enumerate(usernames):
    try:
        print(f"[{index+1}/{len(usernames)}] Fetching details for: {username}")

        # Fetch user details
        user_info = cl.user_info_by_username(username)

        user_details = {
            "username": username,
            "user_id": user_info.pk,
            "full_name": user_info.full_name,
            "followers": user_info.follower_count,
            "following": user_info.following_count,
            "bio": user_info.biography.replace("\n", " "),  # Remove new lines
            "profile_pic_url": user_info.profile_pic_url,
        }

        # Append new data to CSV
        df_output = pd.DataFrame([user_details])
        df_output.to_csv(output_file, index=False, mode="a", header=not os.path.exists(output_file))

        # Wait to prevent rate-limiting
        time.sleep(2)

    except Exception as e:
        print(f"Error fetching {username}: {e}")
        continue

print(f"✅ Process completed. Data saved to {output_file}")
```
