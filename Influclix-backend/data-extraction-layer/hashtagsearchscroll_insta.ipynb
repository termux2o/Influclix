```python
import time
import csv
import os
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# Instagram credentials
INSTAGRAM_USERNAME = "gigi.chauhaan"
INSTAGRAM_PASSWORD = "Tanseem@2004"
HASHTAG = input("Enter hashtag to search for posts (without #): ")  # Instagram hashtag to search
CSV_FILENAME = f"{HASHTAG}.csv"
scroll_wants = int(input("Enter the number of times to scroll: "))

def load_existing_links(filename):
    """Load existing links from the CSV file to avoid duplicates."""
    existing_links = set()
    if os.path.exists(filename):
        with open(filename, mode='r', encoding='utf-8') as file:
            reader = csv.reader(file)
            existing_links = {row[0] for row in reader if row}  # Store existing links in a set
    return existing_links

def instagram_hashtag_posts():
    # Setup WebDriver (Chrome)
    options = webdriver.ChromeOptions()
    options.add_argument("--start-maximized")  # Open browser in maximized mode
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

    try:
        # Load existing links to avoid duplicates
        existing_links = load_existing_links(CSV_FILENAME)

        # Step 1: Log in to Instagram
        print("🔑 Logging into Instagram...")
        driver.get("https://www.instagram.com/accounts/login/")
        time.sleep(5)  # Wait for the login page to load

        # Input username and password
        driver.find_element(By.NAME, "username").send_keys(INSTAGRAM_USERNAME)
        driver.find_element(By.NAME, "password").send_keys(INSTAGRAM_PASSWORD + Keys.RETURN)
        time.sleep(7)  # Wait for login to complete

        # Step 2: Navigate to the hashtag page
        print(f"🔍 Navigating to #{HASHTAG} posts...")
        driver.get(f"https://www.instagram.com/explore/tags/{HASHTAG}/")
        time.sleep(5)  # Wait for the hashtag page to load

        # Step 3: Extract `<a>` tags with href attributes and scroll
        print("📄 Extracting post URLs...")
        extracted_urls = set()  # Store newly extracted URLs

        # Open CSV file in append mode
        with open(CSV_FILENAME, mode='a', newline='', encoding='utf-8') as csvfile:
            csv_writer = csv.writer(csvfile)
            scroll_attempts = 0

            while scroll_attempts < scroll_wants:  # Limit the number of scrolls
                links = driver.find_elements(By.TAG_NAME, "a")
                new_links_found = False

                for link in links:
                    href = link.get_attribute("href")
                    if href and ("/p/" in href or "/reel/" in href) and href not in existing_links and href not in extracted_urls:
                        # Save new URL to the CSV file
                        csv_writer.writerow([href])
                        csvfile.flush()  # Ensure data is written to the file
                        extracted_urls.add(href)  # Add to set to avoid duplicates in the same run
                        new_links_found = True

                if new_links_found:
                    print(f"✅ {len(extracted_urls)} new unique URLs saved...")

                # Scroll down
                print(f"⬇️ Scrolling... (Attempt {scroll_attempts + 1})")
                driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
                time.sleep(1)  # Wait for new content to load

                scroll_attempts += 1

        print(f"\n✅ All extracted links saved to {CSV_FILENAME}.")

    except Exception as e:
        print(f"❌ Error: {e}")

    finally:
        driver.quit()  # Close the browser

# Execute the function
instagram_hashtag_posts()
```

### instaloder bulk extraction usernames by permalinks

```python
import instaloader
import pandas as pd
import csv
import re
import time


INPUT_CSV = "skincarebloggerindia.csv"  # Input file with permalinks
OUTPUT_CSV = "7march.csv"  # Output file
# Load input CSV
df = pd.read_csv(INPUT_CSV)

# Ensure 'Permalink' column exists
if "Permalink" not in df.columns:
    raise ValueError("CSV file must contain a 'Permalink' column.")

# Load existing permalinks from OUTPUT_CSV (to avoid duplicates)
try:
    existing_df = pd.read_csv(OUTPUT_CSV)
    if "Permalink" not in existing_df.columns:
        raise ValueError(f"{OUTPUT_CSV} must contain a 'Permalink' column.")
    existing_permalinks = set(existing_df["Permalink"].astype(str))  # Convert to set for fast lookup
except (FileNotFoundError, pd.errors.EmptyDataError):
    existing_permalinks = set()

def extract_shortcode(permalink):
    """Extracts shortcode from Instagram post URL"""
    match = re.search(r"instagram\.com/p/([A-Za-z0-9_-]+)/", permalink)
    return match.group(1) if match else None

def extract_username(permalink):
    """Extracts username from Instagram post URL structure"""
    match = re.search(r"instagram\.com/p/.*", permalink)
    if match:
        return permalink.split('/')[3]  # Extracting username from URL structure
    return "Unknown"

def get_user_info_from_permalink(permalink):
    """Fetches Instagram username and followers count using Instaloader"""
    try:
        shortcode = extract_shortcode(permalink)
        if not shortcode:
            print(f"❌ Invalid Instagram URL: {permalink}")
            return permalink, extract_username(permalink), "Not Found"

        loader = instaloader.Instaloader()
        post = instaloader.Post.from_shortcode(loader.context, shortcode)
        username = post.owner_username

        try:
            profile = instaloader.Profile.from_username(loader.context, username)
            followers_count = profile.followers
        except Exception:
            followers_count = "Not Found"

        print(f"✅ Extracted: {username} | Followers: {followers_count}")
        return permalink, username, followers_count

    except Exception as e:
        print(f"❌ Error fetching data from {permalink}: {e}")
        return permalink, extract_username(permalink), "Not Found"

# Open output file in append mode
with open(OUTPUT_CSV, mode='a', newline='', encoding='utf-8') as csvfile:
    csv_writer = csv.writer(csvfile)

    # Write headers if file is new
    if not existing_permalinks:
        csv_writer.writerow(["Permalink", "Username", "Followers"])

    # Process each permalink
    for index, row in df.iterrows():
        permalink = str(row["Permalink"]).strip()

        # Skip if already processed
        if permalink in existing_permalinks:
            print(f"⏭️ Skipping (Already Processed): {permalink}")
            continue

        # Fetch data
        permalink, username, followers = get_user_info_from_permalink(permalink)

        # Append to CSV
        csv_writer.writerow([permalink, username, followers])
        csvfile.flush()

        time.sleep(2)  # Increased sleep time to avoid rate limiting

print(f"\n✅ Scraping completed. Data saved to {OUTPUT_CSV}.")
```

### extraction using instaloader after login

```python
import instaloader
import pandas as pd
import csv
import re
import time

INPUT_CSV = "skincarebloggerindia.csv"  # Input file with permalinks
OUTPUT_CSV = "7march.csv"  # Output file

# Prompt for Instagram login credentials
INSTAGRAM_USERNAME = input("Enter Instagram username: ")
INSTAGRAM_PASSWORD = input("Enter Instagram password: ")

# Load input CSV
df = pd.read_csv(INPUT_CSV)

# Ensure 'Permalink' column exists
if "Permalink" not in df.columns:
    raise ValueError("CSV file must contain a 'Permalink' column.")

# Load existing permalinks from OUTPUT_CSV (to avoid duplicates)
try:
    existing_df = pd.read_csv(OUTPUT_CSV)
    if "Permalink" not in existing_df.columns:
        raise ValueError(f"{OUTPUT_CSV} must contain a 'Permalink' column.")
    existing_permalinks = set(existing_df["Permalink"].astype(str))  # Convert to set for fast lookup
except (FileNotFoundError, pd.errors.EmptyDataError):
    existing_permalinks = set()

def extract_shortcode(permalink):
    """Extracts shortcode from Instagram post URL"""
    match = re.search(r"instagram\.com/p/([A-Za-z0-9_-]+)/", permalink)
    return match.group(1) if match else None

def extract_username(permalink):
    """Extracts username from Instagram post URL structure"""
    match = re.search(r"instagram\.com/p/.*", permalink)
    if match:
        return permalink.split('/')[3]  # Extracting username from URL structure
    return "Unknown"

def get_user_info_from_permalink(loader, permalink):
    """Fetches Instagram username and followers count using Instaloader"""
    try:
        shortcode = extract_shortcode(permalink)
        if not shortcode:
            print(f"❌ Invalid Instagram URL: {permalink}")
            return permalink, extract_username(permalink), "Not Found"

        post = instaloader.Post.from_shortcode(loader.context, shortcode)
        username = post.owner_username

        try:
            profile = instaloader.Profile.from_username(loader.context, username)
            followers_count = profile.followers
        except Exception:
            followers_count = "Not Found"

        print(f"✅ Extracted: {username} | Followers: {followers_count}")
        return permalink, username, followers_count

    except Exception as e:
        print(f"❌ Error fetching data from {permalink}: {e}")
        return permalink, extract_username(permalink), "Not Found"

# Initialize Instaloader and login
loader = instaloader.Instaloader()

try:
    loader.login(INSTAGRAM_USERNAME, INSTAGRAM_PASSWORD)
    print("✅ Successfully logged in!")
except Exception as e:
    print(f"❌ Login failed: {e}")
    exit(1)

# Open output file in append mode
with open(OUTPUT_CSV, mode='a', newline='', encoding='utf-8') as csvfile:
    csv_writer = csv.writer(csvfile)

    # Write headers if file is new
    if not existing_permalinks:
        csv_writer.writerow(["Permalink", "Username", "Followers"])

    # Process each permalink
    for index, row in df.iterrows():
        permalink = str(row["Permalink"]).strip()

        # Skip if already processed
        if permalink in existing_permalinks:
            print(f"⏭️ Skipping (Already Processed): {permalink}")
            continue

        # Fetch data
        permalink, username, followers = get_user_info_from_permalink(loader, permalink)

        # Append to CSV
        csv_writer.writerow([permalink, username, followers])
        csvfile.flush()

        time.sleep(2)  # Increased sleep time to avoid rate limiting

print(f"\n✅ Scraping completed. Data saved to {OUTPUT_CSV}.")
```
