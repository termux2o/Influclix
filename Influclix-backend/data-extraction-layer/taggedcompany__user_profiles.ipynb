```python
import pandas as pd
import re
import os

# Input and output file paths
input_csv = "beminimalist__.csv"
output_csv = "users_skinacre.csv"

# Check if the input file exists
if not os.path.exists(input_csv):
    print(f"❌ Error: The input file '{input_csv}' does not exist. Please ensure the file is in the correct directory.")
else:
    try:
        # Load the CSV file
        df = pd.read_csv(input_csv)

        # Ensure 'tagged_links' column exists
        if "tagged_links" not in df.columns:
            raise ValueError("The CSV file does not contain the 'tagged_links' column.")

        # Regular expression to extract usernames
        pattern = r"https://www\.instagram\.com/([^/]+)/"

        # Optional: Process a subset of rows (e.g., from row 964 onward)
        data_r = df["tagged_links"]

        # Extract usernames and create a new DataFrame with only usernames
        usernames = data_r.apply(
            lambda link: re.search(pattern, link).group(1) if pd.notna(link) and re.search(pattern, link) else None
        )
        usernames_df = pd.DataFrame({"usernames": usernames})

        # Save the usernames DataFrame to a new CSV file
        usernames_df.to_csv(output_csv, index=False)
        print(f"✅ Usernames extracted and saved to {output_csv}")

        # Load and display the total number of usernames
        if os.path.exists(output_csv):
            data = pd.read_csv(output_csv)
            print(f"📄 Total unique usernames saved: {len(data)}")
        else:
            print("❌ Error: The output file was not created successfully.")

    except Exception as e:
        print(f"❌ Error: {e}")
```

```python
import pandas as pd
from instascrap import InstaScraper
import os

# Function to fetch Instagram profile details
def fetch_profile_details(username):
    scraper = InstaScraper()
    input_data = {"usernames": [username]}
    results = scraper.Scraper(input_data)
    return results[0] if results else None

# Read usernames from 'usernames.csv'
df_usernames = pd.read_csv('usernames.csv')
usernames_list = df_usernames['usernames'].tolist()

# Define the output CSV file
output_file = 'profiledetails.csv'

# Check if the output file already exists
file_exists = os.path.isfile(output_file)

# Loop through each username and fetch profile details
for username in usernames_list:
    try:
        profile_data = fetch_profile_details(username)
        if profile_data:
            df_profile = pd.DataFrame([profile_data])
            # Append to CSV file
            df_profile.to_csv(output_file, mode='a', header=not file_exists, index=False)
            file_exists = True  # Ensure header is not written again
            print(f"Profile details for '{username}' have been saved.")
        else:
            print(f"No data found for username: {username}")
    except Exception as e:
        print(f"An error occurred while processing username '{username}': {e}")
```

```python
import pandas as pd
import json
import os
from instascrap import InstaScraper

# Function to fetch Instagram profile details
def fetch_profile_details(username):
    scraper = InstaScraper()
    input_data = {"usernames": [username]}
    results = scraper.Scraper(input_data)
    return results[0] if results else None

# Read usernames from 'usernames.csv'
df_usernames = pd.read_csv('usernames.csv')
usernames_list = df_usernames['usernames'].tolist()

# Define the output JSON file
output_file = 'profiledetails.json'

# Check if the output file already exists
if os.path.exists(output_file):
    with open(output_file, 'r', encoding='utf-8') as f:
        try:
            data = json.load(f)
        except json.JSONDecodeError:
            data = []
else:
    data = []

# Loop through each username and fetch profile details
for username in usernames_list:
    try:
        profile_data = fetch_profile_details(username)
        if profile_data:
            data.append(profile_data)
            with open(output_file, 'w', encoding='utf-8') as f:
                json.dump(data, f, indent=4)
            print(f"Profile details for '{username}' have been saved.")
        else:
            print(f"No data found for username: {username}")
    except Exception as e:
        print(f"An error occurred while processing username '{username}': {e}")
```
